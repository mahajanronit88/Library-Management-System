```python
from random import choice
library = []

"""add
   issue
   return
   view
   exit"""

def add_books():
  books = input("Enter a name of a books to be added:")
  library.append(books)
  print(f"{books} has been added")

def issue_books():
  books = input("Enter a name of book to be issued :")

  if books in library:
     library.remove(books)
     print(f"{books} has been removed")
  else:
    print("Book not found")

def return_books():
  books = input("Enter a name of a book to be returned")

  library.append(books)
  print(f"{books} has been returned")

def view_books():
  if not library:
    print("noo books are present in library")

  else:
    print("list of books")
    idx = 1
    for books in library :
      print(f"{idx}.{books}")
      idx +=1
    print()

while(True):
  print("----------Library Management System-----------")

  print("1.Add books")
  print("2.Issue books")
  print("3.Return books")
  print("4.View books")
  print("5.Exit")

  choice = int(input("Enter your choice"))

  if choice==1:
   add_books()
  elif choice==2:
    issue_books()
  elif choice==3:
    return_books()
  elif choice==4:
    view_books()
  elif choice==5:
     print("Thank you, you are now exiting library system")
     break

else :
  print("Invalid choice , Try again !")

```
